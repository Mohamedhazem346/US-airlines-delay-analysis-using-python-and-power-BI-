```python

def convert_time(col):
  
    col = col.fillna(0).astype(int).astype(str).str.zfill(4)
    

    return pd.to_datetime(col, format='%H%M', errors='coerce')

df["DepTime"] = convert_time(df["DepTime"])
df["ArrTime"] = convert_time(df["ArrTime"])
df["CRSDepTime"] = convert_time(df["CRSDepTime"])
df["CRSArrTime"] = convert_time(df["CRSArrTime"])

df[["DepTime", "CRSDepTime", "ArrTime", "CRSArrTime"]].head()
```

Data Cleaning and processing 

```python

essential_cols = ["DepTime", "ArrTime", "ActualElapsedTime", "AirTime", "ArrDelay"]
df = df.dropna(subset=essential_cols) 
df[["TaxiIn", "TaxiOut"]] = df[["TaxiIn", "TaxiOut"]].fillna(0)

delay_cols = ["CarrierDelay", "WeatherDelay", "NASDelay", "SecurityDelay", "LateAircraftDelay"]
df[delay_cols] = df[delay_cols].fillna(0)


df["TailNum"] = df["TailNum"].fillna("Unknown")

df = df.dropna(subset=["CRSArrTime"])


df.isnull().sum()
```

```python
df = df.drop_duplicates()
```
